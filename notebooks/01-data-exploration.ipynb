First, the imports needed

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

Next, define the paths

```python
stops_path = "../data/raw/stops.txt"
sa2_shapes_path = "../data/raw/SA2_2026_AUST_GDA2020.shp"
seifa_path = "../data/raw/SEIFA_2021_IRSAD.csv"
```

Load files. The SEIFA data included 5 rows of header data, so these need to be removed

```python
stops = pd.read_csv(stops_path)
sa2_geo = gpd.read_file(sa2_shapes_path)
seifa = pd.read_csv(
    seifa_path,
    skiprows = 5,
    header = 0)
```

Inspect the shape of the loaded files to check for issues

```python
print("Stops: ", stops.shape)
print("SA2:", sa2_geo.shape)
print("SEIFA:", seifa.shape)
print(stops.columns.to_list())
print(sa2_geo.columns.tolist())
print(seifa.columns.tolist())
```
